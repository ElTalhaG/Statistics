### Example 1: Stochastic variable following a uniform distribution

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

In the standard form, the distribution is uniform on [0, 1]. Using the parameters loc and scale, one obtains the uniform distribution on [loc, loc + scale]. For more, read: https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.uniform.html

```python
# lets plot pdf:
x = np.arange(start=-5, stop=35, step=0.1) # range from -5 to 35
plt.plot(x, stats.uniform.pdf(x, loc=0, scale=30))
plt.show()
```

```python
# lets also plot cdf:
plt.plot(x, stats.uniform.cdf(x, loc=0, scale=30))
plt.show()
```

```python
# Probability P(20 <= X <= 30) = P(X <= 30) - P(X <= 20):
print(stats.uniform.cdf(30, loc=0, scale=30) - stats.uniform.cdf(20, loc=0, scale=30))
```

```python
# visualise the probability:
xint = np.arange(start=20, stop=30, step=0.1) # range from 20 to 30
plt.plot(x, stats.uniform.pdf(x, loc=0, scale=30))
plt.fill_between(xint, stats.uniform.pdf(xint, loc=0, scale=30), color='peachpuff') # color the area
plt.show()
```

```python
# Probability P(x >=30) = 1 - P(X <= 30):
print(1 - stats.uniform.cdf(30, loc=0, scale=30))
```

```python
# visualise the probability
xint = np.arange(start=30, stop=35, step=0.1)
plt.plot(x, stats.uniform.pdf(x, loc=0, scale=30))
plt.fill_between(xint, stats.uniform.pdf(xint, loc=0, scale=30), color='peachpuff')
plt.show()
```

### Example 2: Stochastic variable following a standard normal distribution

A given scale has a measurement error (measured in grams), $Z$, which can be described by a 
standard normal distribution, $ Z \sim N(0,1^2)$.

```python
x = np.arange(start=-5, stop=5, step=0.01) # Again the range by smaller steps
plt.plot(x, stats.norm.pdf(x, loc=0, scale=1)) # loc = mean = mu, scale = standard deviation = sigma
plt.show()
```

a) What is the probability that the scale gives a result that is at least 2 grams less than the true weight of the product?

```python
# Probability P(x <= -2)
print(stats.norm.cdf(x=-2, loc=0, scale=1))

xint = np.arange(start=-5, stop=-2, step=0.01)
plt.plot(x, stats.norm.pdf(x, loc=0, scale=1))
plt.fill_between(xint, stats.norm.pdf(xint, loc=0, scale=1), color='peachpuff')
plt.show()
```

b) What is the probability that the scale gives a result that is at least 2 grams more than the true weight of the product?

```python
# Probability P(x >= 2)
print((1-stats.norm.cdf(2, loc=0, scale=1)))

xint = np.arange(2, 5, .01)
plt.plot(x, stats.norm.pdf(x, loc=0, scale=1))
plt.fill_between(xint, stats.norm.pdf(xint, loc=0, scale=1), color='peachpuff')
plt.show()
```

c) What is the probability that the scale has a deviation of at most $\pm 1$ gram?

```python
# Probability P(-1 <= x <= 1)
print((stats.norm.cdf(1, loc=0, scale=1) - stats.norm.cdf(-1, loc=0, scale=1))*100)

xint = np.arange(-1, 1, .001)
plt.plot(x, stats.norm.pdf(x, loc=0, scale=1))
plt.fill_between(xint, stats.norm.pdf(xint, loc=0, scale=1), color='peachpuff')
plt.show()
```

### Example 3: Stochastic variable following a normal distribution

It is assumed that the salary of primary school teachers can be described by a normal distribution with <br>
a mean value of $\mu=290$ (in 1000 DKK) and a standard deviation of $\sigma=4$ (1000 DKK).

```python
x = np.arange(270, 310, .1)
plt.plot(x, stats.norm.pdf(x, loc=290, scale=4))
plt.show()
```

a) What is the probability that a randomly selected teacher earns more than 300,000 DKK?

```python
# Probability P(x > 300)
print(1 - stats.norm.cdf(x=300, loc=290, scale=4))

xint = np.arange(300, 310, .1)
plt.plot(x, stats.norm.pdf(x, loc=290, scale=4))
plt.fill_between(xint, stats.norm.pdf(xint, loc=290, scale=4), color='peachpuff')
plt.show()
```

b) (Inverse question) Specify a salary range (which is symmetric around the mean) that covers 95\% of teachers' salaries.

```python
# 95% interval around mean = 290

# we need 2.5% on either side of the interval

# lower bound should be where cdf = 0.025
# upper bound should be where cdf = 0.975

plt.plot(x,stats.norm.cdf(x, loc=290, scale=4))
plt.axhline(0.025, linestyle='--', color="black", xmin=0, xmax=.32)
plt.axhline(0.975, linestyle='--', color="black", xmin=0, xmax=(1-.32))
plt.xlabel("x")
plt.ylabel("cdf(x)")
plt.show()
```

To find the corresponding values of x we need the inverse of cdf(x)

```python

# in Python (Scipy.stats package) the inverse of "cdf" is called "ppf" ("percent point function") - The quantile function (percentile)
# scale is the standard deviation, 

# lower bound where cdf = 0.025:
lower_bound = stats.norm.ppf(0.025, loc=290, scale=4)
print(lower_bound)
```

```python

# upper bound where cdf = 0.975:
upper_bound = stats.norm.ppf(0.975, loc=290, scale=4)
print(upper_bound)
```

```python
# lets also plot these values:
plt.plot(x,stats.norm.cdf(x, loc=290, scale=4))
plt.axhline(0.025, linestyle='--', color="black", xmin=0, xmax=.32)
plt.axhline(0.975, linestyle='--', color="black", xmin=0, xmax=(1-.32))
plt.axvline(lower_bound, linestyle='--', color="black", ymin=0, ymax=.07)
plt.axvline(upper_bound, linestyle='--', color="black", ymin=0, ymax=1-.07)
plt.show()
```

```python
# lets also plot pdf with interval:
xint = np.arange(lower_bound, upper_bound, .1)
plt.plot(x, stats.norm.pdf(x, loc=290, scale=4))
plt.fill_between(xint, stats.norm.pdf(xint, loc=290, scale=4), color='peachpuff')
plt.show()
```

### Example 4: Stochastic variable following an exponential distribution

The time between customer arrivals at a post office follows an exponential distribution with a mean of $\mu = 2$ minutes.

```python
# X ~ exp(lambda)

# mean = 2
# lambda = 1/2 = 0.5

x = np.arange(start=0, stop=5, step=0.01)
plt.plot(x, stats.expon.pdf(x, loc=0, scale=2)) #scale= 1/lambda = mean
plt.show()
```

a) A customer has just arrived. What is the probability that no more customers will arrive within a period of 2 minutes?

```python
# X ~ exp(lambda)
# a) Probability P(X > 2)
print(1 - stats.expon.cdf(x=2, loc=0, scale=2))
```

```python
# lets also plot this probability:
xint = np.arange(2, 5, .01)
plt.plot(x, stats.expon.pdf(x, loc=0, scale=2))
plt.fill_between(xint, stats.expon.pdf(xint, loc=0, scale=2), color='peachpuff')
plt.show()
```

#### b) Now use the Poisson distribution to calculate the probability that no more customers will arrive within the next two minutes.

Poisson - count of events per time-interval.

We want zero events (zero customers) in a time interval of 2 minutes. Average waiting time is 2 minutes, so the average count per two-minute interval (average rate) is 1

Now, we use the poisson distribution with fixed time interval of 2 minutes and calculate the probability of zero events P(X=0)

```python
# b) Using poissondistribution to answer the same question

print(stats.poisson.pmf(k=0,mu=1))
```

```python
# we can also plot the pdf for number of customers within next two minutes:
plt.bar(np.arange(start=0,stop=10,step=1), stats.poisson.pmf(np.arange(0,10,1), 1), width=0.1, color="red")
plt.show()
```

To simulate "random variates" we use the .rvs method

### Example 6: Rules for stochastic variables 

```python
# simulate 1000 realisations of the random variable X:
x = stats.norm.rvs(loc=4, scale=np.sqrt(6), size=1000)
plt.hist(x, density=True)
plt.plot(np.arange(4-20,4+20,0.01), stats.norm.pdf(np.arange(4-20,4+20,0.01), loc=4,scale=np.sqrt(6)), color="red")
plt.xlim(-40,30)
plt.show()
```

Calculate the mean and variance values of the stochastic variable Y:

$Y = -3X + 2$

$E[Y] = E[-3X+2] = -3E[X] + 2 = -3\cdot 4 + 2 = -10$

$V[Y] = V[-3X + 2] = (-3)^2V[X] = 9\cdot 6 = 54$

```python
# Visualizaing variable x and variable y = -3x + 2:
y = -3*x + 2

plt.hist(x, density=True)
plt.plot(np.arange(4-20,4+20,0.01), stats.norm.pdf(np.arange(4-20,4+20,0.01), loc=4,scale=np.sqrt(6)), color="red")

plt.hist(y, density=True)
plt.plot(np.arange(4-50,4+50,0.01), stats.norm.pdf(np.arange(4-50,4+50,0.01), loc=-10,scale=np.sqrt(54)), color="red")
plt.xlim(-40,30)
plt.show()
```
